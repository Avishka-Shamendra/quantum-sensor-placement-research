# Extended ablation study

Allocation and test-order comparisons on the completed extended datasets. The curves give 1D and 2D equal weight. Each instance and budget uses one seeded random tree.


```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
from IPython.display import display

working = Path.cwd().resolve()
PROJECT = working / "problem-2-min-error-seq" if (working / "problem-2-min-error-seq").is_dir() else working
SOURCE = PROJECT / "src"
if str(SOURCE) not in sys.path:
    sys.path.insert(0, str(SOURCE))

from ablation import run_ablation, load_ablation, plot_ablation
```

## Settings

Run notebook 02 first with the same mode and no time limit. Saved Greedy and VND allocations and VND + Beam Search errors are reused. Additional trees use the same beam width and positive-sensor location rule.


```python
RUN_MODE = "full"
RUN_ANALYSIS = True
RESUME = True
```

## Matched comparisons

Checkpoints retain each completed instance-budget comparison. No datasets are generated and no time limits are imposed.


```python
if RUN_ANALYSIS:
    analysis = run_ablation(project=PROJECT, mode=RUN_MODE, resume=RESUME)
else:
    analysis = load_ablation(project=PROJECT, mode=RUN_MODE)

print("Results:", analysis["output"])
```

## Ablation figure

Left: compare sensor allocations with Beam Search fixed. Right: compare one random tree with Beam Search for the same VND allocation.


```python
figure = plot_ablation(analysis["summary"])
output = analysis["output"] / "figures"
output.mkdir(parents=True, exist_ok=True)
for extension in ("pdf", "png"):
    figure.savefig(output / f"extended_ablation.{extension}", dpi=300, bbox_inches="tight")
display(figure)
plt.close(figure)
```
